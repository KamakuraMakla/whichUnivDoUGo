```python
import pandas as pd
from pathlib import Path
%pip install xlrd
```

```python
data_dir = Path('../data/raw/school_census')

for year in [1975, 1995, 2015, 2025]:
    files = list(data_dir.glob(f'{year}.xls*'))

    if not files:
        print(f'{year}: ファイルなし')
        continue

    file = files[0]
    xls = pd.ExcelFile(file)

    print(f'\n===={year}年====')
    print("シート:", xls.sheet_names)

    df = pd.read_excel(file, sheet_name=0, header=None)
```

```python
for year in [1975, 1995, 2015, 2025]:
    file = list(data_dir.glob(f'{year}.xls*'))[0]
    xls = pd.ExcelFile(file)

    print(f'\n==== {year}年 ====')

    for sheet in xls.sheet_names:
        df = pd.read_excel(xls, sheet_name=sheet, header=None)
        print(f'{sheet}: {df.shape}')
```

```python
for year in [1975, 2025]:
    file = list(data_dir.glob(f'{year}.xls*'))[0]
    xls = pd.ExcelFile(file)

    for sheet in [xls.sheet_names[0], xls.sheet_names[-1]]:
        df = pd.read_excel(xls, sheet_name=sheet, header=None)

        print(f'\n==== {year}年 / {sheet} ====')
        print(df.iloc[:15, :15].to_string())
```
